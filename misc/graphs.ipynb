```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

baseline_cm = np.array([
    [2490, 367, 102, 41],
    [324, 2556, 65, 55],
    [88, 69, 323, 20],
    [75, 68, 29, 328]
])

lnr_cm = np.array([
    [2488, 367, 104, 41],
    [321, 2555, 67, 57],
    [86, 69, 325, 20],
    [74, 69, 29, 328]
])

sns.set_theme(style="darkgrid")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

labels = ['4/4', '3/4', '5/4', '7/4']

sns.heatmap(baseline_cm, ax=axes[0], annot=True, fmt='d', cmap='Blues', 
            xticklabels=labels, yticklabels=labels, cbar=False, annot_kws={"size": 16})
axes[0].set_title('Baseline Time Signature Confusion Matrix', fontsize=20)
axes[0].set_ylabel('True Label')
axes[0].set_xlabel('Predicted Label')

sns.heatmap(lnr_cm, ax=axes[1], annot=True, fmt='d', cmap='Blues', 
            xticklabels=labels, yticklabels=labels, cbar=False, annot_kws={"size": 16})
axes[1].set_title('LNR Time Signature Confusion Matrix', fontsize=20)
axes[1].set_ylabel('True Label')
axes[1].set_xlabel('Predicted Label')

plt.tight_layout()
plt.show()
```
