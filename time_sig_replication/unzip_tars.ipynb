## unzip datasets

```python
dataset_folder = "./dataverse_files"
```

```python

FMA_dataset = "FMA.tar.gz"
# unzip FMA
import tarfile
with tarfile.open(f"{dataset_folder}/{FMA_dataset}", "r:gz") as tar:
    tar.extractall(path=dataset_folder)

```
