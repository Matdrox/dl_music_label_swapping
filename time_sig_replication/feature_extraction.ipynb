```python
# Function to each 30 second file into 3 second segments, keep sorted by file
def segment_audio_files(audio_files, segment_duration=3, sr=22050):
    segmented_files = {}
    for file_path, y in audio_files:
        total_segments = len(y) // (segment_duration * sr)
        segments = []
        for i in range(total_segments):
            start_sample = i * segment_duration * sr
            end_sample = start_sample + segment_duration * sr
            segment = y[start_sample:end_sample]
            print(len(segment)) # 66150... 
            segments.append(segment) 
        segmented_files[file_path] = segments
    return segmented_files


# use librosa.util.frame instead
def segment_audio_files(audio_files, segment_duration=3, sr=22050, hop_length=512):
    segmented_files = {}
    for file_path, y in audio_files:
        frame_length = segment_duration * sr
        frames = librosa.util.frame(y, frame_length=frame_length, hop_length=hop_length).T
        segments = [frame for frame in frames]
        segmented_files[file_path] = segments
    return segmented_files


def segment_audio(y, sr, segment_seconds=3, hop_length=512):
    frame_length = int(segment_seconds * sr)

    # frames is a 2D view: shape = (frame_length, n_frames)
    frames = librosa.util.frame(y, frame_length=frame_length, hop_length=hop_length)
    
    # return shape transposed → (n_frames, frame_length)
    return frames.T


""" print(len(OWN_audio_files)) # prints 113 - there are only 113 files
OWN_segmented_files = segment_audio_files(OWN_audio_files)
print(OWN_segmented_files)

print('')

for key in OWN_segmented_files:
    print(f"{key}: {len(OWN_segmented_files[key])} segments")
    #print(len(OWN_segmented_files[key][0]))  # prints length of first segment to verify

plot_waveform(OWN_segmented_files[list(OWN_segmented_files.keys())[0]][0])
"""



```

```python
def compute_mfcc_segments(segments, sr):
    # segments: shape (n_segments, segment_length)
    mfcc_list = []
    for seg in segments:
        mfcc = librosa.feature.mfcc(
            y=seg,
            sr=sr,
            n_fft=2048,
            hop_length=512,
            n_mels=128,
            n_mfcc=20
        )
        mfcc_list.append(mfcc)
    return mfcc_list
```
